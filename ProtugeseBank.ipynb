```python
# Create Dataset for Machine Learning

model_df = df.drop(
    columns=['pdays_group', 'duration_group', 'y_encoded', 'duration']
)

print("Modeling Dataset Shape:", model_df.shape)

print("\nRemaining Columns:")
print(model_df.columns.tolist())

print("\nTarget Distribution:")
display(model_df['y'].value_counts())
```

```python
# Separate Input Features and Target Variable

X = model_df.drop(columns=['y'])

y = model_df['y'].map({
    'no': 0,
    'yes': 1
})

print("Input Features Shape:", X.shape)
print("Target Variable Shape:", y.shape)

print("\nTarget Class Distribution:")
print(y.value_counts())
```

```python
# Identify Numerical and Categorical Features

numerical_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object', 'str']
).columns.tolist()

print("Numerical Features:", len(numerical_features))
print(numerical_features)

print("Categorical Features:", len(categorical_features))
print(categorical_features)
```

```python
# Split Dataset into Training and Testing Sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("Testing Target Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
# Encode Categorical Features

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(
    X_train[categorical_features]
)

X_test_encoded = encoder.transform(
    X_test[categorical_features]
)

print("Training Encoded Shape:", X_train_encoded.shape)
print("Testing Encoded Shape:", X_test_encoded.shape)
```

```python
# Scale Numerical Features

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train[numerical_features]
)

X_test_scaled = scaler.transform(
    X_test[numerical_features]
)

print("Training Scaled Shape:", X_train_scaled.shape)
print("Testing Scaled Shape:", X_test_scaled.shape)
```

```python
# Combine Numerical and Categorical Features

X_train_final = np.hstack([
    X_train_scaled,
    X_train_encoded
])

X_test_final = np.hstack([
    X_test_scaled,
    X_test_encoded
])

print("Final Training Shape:", X_train_final.shape)
print("Final Testing Shape:", X_test_final.shape)
```

```python
# Train Logistic Regression Model

from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

lr_model.fit(X_train_final, y_train)

print("Logistic Regression Model Trained Successfully!")
```

```python
# Make Predictions Using Logistic Regression

y_pred_lr = lr_model.predict(X_test_final)

print("First 20 Actual Values:")
print(y_test.values[:20])

print("First 20 Predicted Values:")
print(y_pred_lr[:20])
```

```python
# Evaluate Logistic Regression Model

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

y_prob_lr = lr_model.predict_proba(X_test_final)[:, 1]

print("Logistic Regression Performance")
print("--------------------------------")
print("Accuracy :", round(accuracy_score(y_test, y_pred_lr), 4))
print("Precision:", round(precision_score(y_test, y_pred_lr), 4))
print("Recall   :", round(recall_score(y_test, y_pred_lr), 4))
print("F1 Score :", round(f1_score(y_test, y_pred_lr), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob_lr), 4))
```

```python
# Visualize Logistic Regression Confusion Matrix

from sklearn.metrics import confusion_matrix

cm_lr = confusion_matrix(y_test, y_pred_lr)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm_lr,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['No', 'Yes'],
    yticklabels=['No', 'Yes']
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# Generate Logistic Regression Classification Report

from sklearn.metrics import classification_report

print("Logistic Regression Classification Report")
print("------------------------------------------")

print(
    classification_report(
        y_test,
        y_pred_lr,
        target_names=['No Subscription', 'Yes Subscription']
    )
)
```
